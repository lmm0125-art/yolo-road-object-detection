```python
#1
import torch

x = torch.tensor(2.0, requires_grad=True)
y = x**2

y.backward()

print(x.grad)
```

```python
#2
x = torch.tensor(2.0, requires_grad=True)

y = x**2
y.backward()
print("첫 번째 계산 후 x.grad (dy/dx):", x.grad)
z = x**3
z.backward()
print("두 번째 계산 후 누적된 x.grad (dy/dx + dz/dx):", x.grad)
```

```python
#3
x = torch.tensor(2.0, requires_grad=True)
y = x**2
y.backward()
x.grad.zero_()
print("초기화 후 x.grad:", x.grad)
z = x**3
z.backward()
print("다시 계산된 x.grad (dz/dx):", x.grad)
```

```python
#4
x = torch.tensor(2.0, requires_grad=True)
y = x**2
y.retain_grad()
z = y**2
z.backward()
print("y의 기울기 (dz/dy):", y.grad)
print("x의 기울기 (dz/dx):", x.grad)
```

```python
#5
x = torch.tensor(2.0, requires_grad=True)
y = x**2
y_np = y.detach().numpy()
print(type(y_np))
print("Numpy array:", y_np)
```

```python
#6
def f(x):
    return x**2
def diff(f, x, h=1e-4):
    return (f(x + h) - f(x - h)) / (2 * h)
result = diff(f, 2.0)
print(f"x=2.0 에서의 수치 미분값: {result}")
```

```python
#7
x = torch.tensor(2.0, requires_grad=True)
y = torch.tensor(3.0, requires_grad=True)
z = x**2 + y**2
z.backward()

print("x의 기울기 (dz/dx):", x.grad)
print("y의 기울기 (dz/dy):", y.grad)
```

```python
#8
x = torch.tensor([2.0, 3.0, 4.0], requires_grad=True)
y = x**2
y.backward(torch.ones_like(x))

print("x의 기울기:", x.grad)
```

```python
#9
x = torch.tensor(2.0, requires_grad=True)
y = torch.tensor(3.0, requires_grad=True)
z = x**2 + y**2
grads = torch.autograd.grad(z, (x, y))

print("x의 기울기 (dz/dx):", grads[0])
print("y의 기울기 (dz/dy):", grads[1])
```

```python
#10
import torch
from torch.autograd.functional import jacobian

def f(x):
    return x**2

x = torch.tensor([2.0, 3.0, 4.0])
J = jacobian(f, x)
print("Jacobian 행렬:\n", J)
```
