```python
#1
import torch

x = torch.arange(12)
y = x.view(3, 4) # 또는 x.reshape(3, 4)
print(y.shape)
```

```python
#2（1）변화: y와 z의 첫 번째 요소(y[0, 0] 및 z[0, 0, 0]) 역시 모두 100으로 변경됩니다.
#（2）이유: view() 함수를 통해 생성된 텐서들은 새로운 메모리를 할당받지 않고 원본 텐서 x와 동일한 메모리를 공유하기 때문입니다.
```

```python
#3
x = torch.arange(10)
print(x.is_contiguous())
#is_contiguous()의 의미: 텐서의 데이터가 메모리 상에 순차적(연속적)으로 배열되어 있는지를 확인하여, 연속적일 경우 True를 반환합니다.
```

```python
#6
x = torch.tensor([[1, 2, 3], [4, 5, 6]])
y = x.transpose(0, 1)
print(y.shape)
```

```python
#7（1） 오류 원인: transpose() 연산은 텐서의 형태(뷰)만 바꾸기 때문에 메모리 구조가 비연속적(non-contiguous) 상태가 됩니다. view() 함수는 메모리가 연속적인 텐서에만 적용 가능하므로 오류가 발생합니다.
#（2）
x = torch.arange(6).view(2, 3)
y = x.transpose(0, 1)
z = y.contiguous().view(6)
```

```python
#8（1）
x = torch.arange(6).view(1, 6, 1)
y = x.squeeze()
#（2）
z = y.unsqueeze(0).unsqueeze(2)
```

```python
#9
x = torch.tensor([[1, 2], [3, 4]])
y = torch.tensor([[5, 6]])
z1 = torch.cat([x, y], dim=0)
z2 = torch.concat([x, y], dim=0)
```

```python
#10（1）
x = torch.tensor([[1, 2], [3, 4]])
y1 = torch.stack([x, x, x], dim=0)
#（2）
y2 = x.unsqueeze(0).expand(3, 2, 2)
#（3）
y3 = x.unsqueeze(0).repeat(3, 1, 1)
#메모리 사용 차이 설명: expand()는 실제로 데이터를 복사하지 않고 기존 메모리의 뷰(view)만 확장하여 보여주기 때문에 메모리를 절약할 수 있습니다. 반면, repeat()는 텐서의 데이터를 실제로 새로운 메모리 공간에 복사하여 반복하므로 메모리 사용량이 크게 증가합니다.
```
