```python
#1
import torch

x = torch.arange(1, 26).reshape(5, 5)
print(x[0, :])
print(x[-1, :])
print(x[:, 0])
print(x[:, -1])
```

```python
#2
x = torch.arange(20).reshape(4, 5)
y = x[:, [4, 2, 0]]
print(y)
```

```python
#3
x = torch.arange(1, 17).reshape(4, 4)
print(x[:2, :2])
print(x[:2, 2:])
print(x[2:, :2])
print(x[2:, 2:])
```

```python
#4
y = torch.arange(24).reshape(2, 3, 4)
print(y[0, 0, 0])
print(y[:, :, 0])
print(y[:, :, 1])
print(y[:, :, 0])
print(y[..., 0])
```

```python
#5
scores = torch.tensor([[70, 80, 90], [60, 75, 85], [90, 88, 92]])
print(scores[0, :])
print(scores[:, 1])
print(scores[:, ::2])
```

```python
#6
z = torch.arange(-10, 10)
print(z > 0)
print(z[z > 0])
z[z < 0] = 0
print(z)
```

```python
#7
a = torch.tensor([True, False, True, False, True])
b = torch.tensor([True, True, False, False, True])
print(torch.logical_and(a, b))
print(torch.logical_or(a, b))
print(torch.logical_not(a))
```

```python
#8
x = torch.arange(36).reshape(6, 6)
print(x[::2, :])
print(x[:, 1::2])
print(x[::2, 1::2])
```

```python
#9
t = torch.arange(2 * 4 * 5).reshape(2, 4, 5)
print(t[0, :, :])
print(t[1, :, :])
print(t[:, 0, :])
print(t[:, :, -1])
```

```python
#10
x = torch.tensor([True, True, False, False])
y = torch.tensor([True, False, True, False])
print(x & y)
print(~x)
```
