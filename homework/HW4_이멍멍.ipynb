```python
#1
import torch
x1 = torch.tensor([[3, 7, 2],
                   [9, 1, 5],
                   [4, 8, 6]])

min_val = x1.min()
max_val = x1.max()
print(" 전체 최솟값:", min_val.item())
print(" 전체 최댓값:", max_val.item())
```

```python
#2
x2 = torch.tensor([[5, 8, 3, 7],
                   [2, 6, 9, 1],
                   [4, 0, 6, 3]])

min_vals, min_indices = x2.min(dim=0)
print("dim=0 최솟값:", min_vals)
print("인덱스:", min_indices)
```

```python
#3
x3 = torch.tensor([[1, 9, 4, 2],
                   [8, 3, 7, 5],
                   [6, 0, 2, 11]])

max_vals, max_indices = x3.max(dim=1)
print("dim=1 최댓값:", max_vals)
print("인덱스:", max_indices)
```

```python
#4
x4 = torch.tensor([[10, 4, 7, 3],
                   [8, 15, 1, 6],
                   [9, 2, 5, 12]])

flat_argmin = x4.argmin()
flat_argmax = x4.argmax()

rows, cols = x4.shape
min_pos = (flat_argmin // cols, flat_argmin % cols)
max_pos = (flat_argmax // cols, flat_argmax % cols)

print("전체 최솟값 위치 (flatten 인덱스):", flat_argmin.item())
print("2차원 좌표:", min_pos)
print("전체 최댓값 위치 (flatten 인덱스):", flat_argmax.item())
print("2차원 좌표:", max_pos)
```

```python
#5
x5 = torch.arange(1, 7).reshape(2, 3)
print("원본 텐서:")
print(x5)

total_sum = x5.sum()
row_sum = x5.sum(dim=1)
col_sum = x5.sum(dim=0)

print("전체 합:", total_sum.item())
print("행별 합:", row_sum)
print("열별 합:", col_sum)
```

```python
#6
x6 = torch.tensor([[1.0, 2.5, 3.5],
                   [4.0, 5.5, 6.5]])

mean6 = x6.mean()
var6 = x6.var()
std6 = x6.std()

print("전체 평균:", mean6.item())
print("전체 분산:", var6.item())
print("전체 표준편차:", std6.item())
```

```python
#7
x7 = torch.tensor([-4, 7, -2, 9, -1, 3], dtype=torch.float)
result7 = torch.where(x7 > 0, x7, torch.tensor(0.0))
print("torch.where 결과:", result7)
```

```python
#8
x8 = torch.tensor([7, 2, 9, 1, 5, 3])
sorted_vals, indices = torch.sort(x8)
argsort_result = x8.argsort()

print("정렬된 값:", sorted_vals)
print("정렬된 인덱스 (torch.sort):", indices)
print("torch.argsort 결과:", argsort_result)
```

```python
#9
A = torch.tensor([[1, 2],
                  [3, 4]], dtype=torch.float)
B = torch.tensor([[5, 6],
                  [7, 8]], dtype=torch.float)

matmul_result = torch.matmul(A, B)
at_result = A @ B
are_equal = torch.allclose(matmul_result, at_result)

print("torch.matmul 결과:\n", matmul_result)
print("@ 연산자 결과:\n", at_result)
print("두 결과가 같은가?", are_equal)
```

```python
#10
A10 = torch.tensor([[2.0, 1.0],
                    [5.0, 3.0]])
b10 = torch.tensor([1.0, 2.0])

x10 = torch.linalg.solve(A10, b10)
print("선형방정식의 해:", x10)
```
